```python
# ============================================================
# Recomendador de Bases via Fatoração Prima
# ============================================================
# Duas famílias de bases, separadas por um muro:
#   - Bases potência de 2:  2, 4, 8, 16, 32, 64, ...
#   - Bases com primos 3 e 5: 10, 12, 60, 360, ...
# Dentro da família: conversão livre (reagrupamento).
# Entre famílias: aritmética obrigatória.

def fatorar(n):
    """Fatoração prima de n."""
    fatores = {}
    d = 2
    while d * d <= n:
        while n % d == 0:
            fatores[d] = fatores.get(d, 0) + 1
            n //= d
        d += 1
    if n > 1:
        fatores[n] = fatores.get(n, 0) + 1
    return fatores


def primos(base):
    """Conjunto de primos na fatoração da base."""
    return set(fatorar(base).keys())


def custo_conversao(base_a, base_b):
    """
    Classificação correta:
      'livre'      - mesmos primos (reagrupamento puro)
      'aritmetico' - primos diferentes (multiplicação/divisão)
    """
    if primos(base_a) == primos(base_b):
        return 'livre'
    return 'aritmetico'


def analisar_bases(bases):
    """Raio-X: quais pares convertem livremente, quais exigem aritmética."""
    print(f"{'':>6}", end="")
    for b in bases:
        print(f"{b:>6}", end="")
    print()
    for a in bases:
        print(f"{a:>6}", end="")
        for b in bases:
            c = custo_conversao(a, b)
            simbolo = {'livre': '·', 'aritmetico': '#'}[c]
            print(f"{simbolo:>6}", end="")
        print()
    print("\nLegenda: · livre   # aritmético")


def recomendar_bases(bases_alvo, bases_candidatas=None):
    """
    Dado um conjunto de bases que precisamos suportar,
    sugere o pivô que maximiza conversões livres.
    """
    if bases_candidatas is None:
        bases_candidatas = [
            2, 4, 8, 16, 32, 64, 128,       # família binária
            10, 100, 1000,                   # família decimal
            12, 24, 60, 120, 360,            # bases compostas
        ]

    resultados = []
    for pivo in bases_candidatas:
        livres = 0
        aritmeticas = 0
        for base in bases_alvo:
            c = custo_conversao(pivo, base)
            if c == 'livre':
                livres += 1
            else:
                aritmeticas += 1
        # Ordena por: mais livres primeiro, menos aritméticas depois
        resultados.append((pivo, livres, aritmeticas))

    resultados.sort(key=lambda x: (-x[1], x[2], x[0]))
    return resultados


def cenarios():
    """Demonstra o recomendador em casos práticos."""
    print("=" * 60)
    print("RECOMENDADOR DE BASES VIA FATORAÇÃO PRIMA")
    print("=" * 60)

    print("\n--- Mapa de relações entre bases comuns ---")
    analisar_bases([2, 8, 16, 32, 10, 12, 60])

    cenarios_teste = [
        ("Só binário/octal/hexa (memória)",
         [2, 8, 16]),
        ("Binário + decimal (sistemas comuns)",
         [2, 10]),
        ("Decimal + tempo (base 60)",
         [10, 60]),
        ("Hexadecimal + decimal + binário (protocolos)",
         [2, 10, 16]),
        ("Todas as famílias (JSON + IP + permissões Unix + TOTP)",
         [10, 16, 8, 32]),
    ]

    for nome, bases in cenarios_teste:
        print(f"\n--- {nome} ---")
        print(f"Bases alvo: {bases}")
        resultados = recomendar_bases(bases)
        for pivo, livres, art in resultados[:3]:
            print(f"  pivô {pivo:>5}: {livres} livres, {art} aritméticas")


if __name__ == "__main__":
    cenarios()
```

```python
# ============================================================
# SRCC-Decision: Recomendador de Bases com Custo Ponderado
# ============================================================
# Extensão do recomendador anterior:
#   - Custo numérico por primo (não apenas "livre vs pago")
#   - Workload com frequências (peso por base usada)
#   - Análise de custo amortizado
#   - Análise de sensibilidade (como a recomendação muda com a carga)
#   - Visualização gráfica

from collections import defaultdict
import matplotlib.pyplot as plt

# ------------------------------------------------------------
# 1. Custo por primo
# ------------------------------------------------------------

CUSTO_PRIMO = {
    2: 1,      # shift / mask, quase grátis
    3: 3,      # multiply leve
    5: 3,      # multiply leve
    7: 7,      # multiply
    11: 11, 13: 13, 17: 17, 19: 19,
}

def custo_primo(p):
    """Custo estimado de multiplicar/dividir por p."""
    return CUSTO_PRIMO.get(p, p)

def fatorar(n):
    """Fatoração prima de n."""
    fatores = {}
    d = 2
    while d * d <= n:
        while n % d == 0:
            fatores[d] = fatores.get(d, 0) + 1
            n //= d
        d += 1
    if n > 1:
        fatores[n] = fatores.get(n, 0) + 1
    return fatores

# ------------------------------------------------------------
# 2. Custo de conversão entre duas bases
# ------------------------------------------------------------

def custo_conversao(base_a, base_b):
    """
    Custo de converter um número representado em base_a para base_b.

    Modelo: o custo é a soma ponderada dos primos cujos expoentes
    diferem entre as duas bases, multiplicada pela magnitude da diferença.

    Propriedades:
      - Simétrico: custo(A,B) == custo(B,A)
      - Zero se A == B
      - Zero se A e B têm exatamente o mesmo conjunto de primos
    """
    if base_a == base_b:
        return 0
    fa = fatorar(base_a)
    fb = fatorar(base_b)
    todos = set(fa) | set(fb)
    custo = 0
    for p in todos:
        delta = fb.get(p, 0) - fa.get(p, 0)
        if delta != 0:
            custo += abs(delta) * custo_primo(p)
    return custo

# ------------------------------------------------------------
# 3. Análise amortizada
# ------------------------------------------------------------

def custo_amortizado(pivo, workload):
    """
    workload: lista de (base, peso), onde peso representa a frequência
    relativa daquela base no sistema.

    Retorna o custo total de usar `pivo` como base intermediária
    em todas as conversões.
    """
    return sum(peso * custo_conversao(pivo, base) for base, peso in workload)

def normalizar_workload(workload):
    """Normaliza os pesos para somarem 1.0."""
    total = sum(p for _, p in workload)
    if total == 0:
        return workload
    return [(b, p / total) for b, p in workload]

def recomendar_pivo(workload, candidatos=None):
    """
    Retorna a lista de pivôs ordenada por custo amortizado crescente.
    """
    if candidatos is None:
        candidatos = [2, 4, 8, 16, 32, 64, 128,
                      10, 100, 1000,
                      12, 24, 60, 120, 360]

    resultados = []
    for pivo in candidatos:
        custo = custo_amortizado(pivo, workload)
        resultados.append((pivo, custo))

    resultados.sort(key=lambda x: x[1])
    return resultados

# ------------------------------------------------------------
# 4. Análise de sensibilidade
# ------------------------------------------------------------

def analise_sensibilidade(workload_base, base_variavel,
                          pesos=(0.0, 0.1, 0.25, 0.5, 0.75, 0.9, 1.0),
                          candidatos=None):
    """
    Mantém todas as bases fixas exceto uma, que tem seu peso variado.
    Retorna uma tabela mostrando como o pivô ótimo muda.
    """
    if candidatos is None:
        candidatos = [2, 8, 16, 10, 60]

    # Bases fixas (excluindo a variável)
    bases_fixas = [(b, p) for b, p in workload_base if b != base_variavel]
    peso_fixo_total = sum(p for _, p in bases_fixas)

    resultados = []
    for peso_var in pesos:
        # Constrói workload com peso normalizado
        restante = 1.0 - peso_var
        if peso_fixo_total > 0:
            workload = [(b, p * restante / peso_fixo_total) for b, p in bases_fixas]
        else:
            workload = []
        workload.append((base_variavel, peso_var))

        ranking = recomendar_pivo(workload, candidatos)
        melhor_pivo, melhor_custo = ranking[0]
        resultados.append((peso_var, melhor_pivo, melhor_custo, ranking[:3]))

    return resultados

# ------------------------------------------------------------
# 5. Visualização
# ------------------------------------------------------------

def plotar_custos(workload, candidatos=None, titulo="Custo por pivô"):
    """Gera gráfico de barras do custo amortizado por pivô."""
    if candidatos is None:
        candidatos = [2, 8, 16, 10, 60]

    custos = [(p, custo_amortizado(p, workload)) for p in candidatos]
    custos.sort(key=lambda x: x[1])

    nomes = [str(p) for p, _ in custos]
    valores = [c for _, c in custos]

    plt.figure(figsize=(10, 5))
    barras = plt.bar(nomes, valores, color='steelblue')
    # Destaca o melhor
    barras[0].set_color('darkgreen')
    plt.xlabel("Pivô")
    plt.ylabel("Custo amortizado")
    plt.title(titulo)
    plt.grid(axis='y', alpha=0.3)
    plt.tight_layout()
    plt.show()
    return custos

# ------------------------------------------------------------
# 6. Cenários práticos
# ------------------------------------------------------------

def cenario(nome, workload, candidatos=None):
    """Executa e imprime o resultado de um cenário."""
    print("\n" + "=" * 65)
    print(f"CENÁRIO: {nome}")
    print("=" * 65)
    workload_norm = normalizar_workload(workload)
    print("Workload (normalizado):")
    for b, p in workload_norm:
        print(f"  base {b:>5}  peso {p:.3f}")

    ranking = recomendar_pivo(workload_norm, candidatos)
    print("\nTop 5 pivôs:")
    for i, (pivo, custo) in enumerate(ranking[:5], 1):
        marcador = " ◀ ótimo" if i == 1 else ""
        print(f"  {i}. pivô {pivo:>5}  →  custo {custo:.4f}{marcador}")

    return ranking

# ------------------------------------------------------------
# 7. Execução
# ------------------------------------------------------------

if __name__ == "__main__":
    print("=" * 65)
    print("SRCC-Decision: Recomendador de Bases com Custo Ponderado")
    print("=" * 65)

    # --- Cenário A: ambiente puramente binário
    cenario("Ambiente binário puro (memória, redes)",
            [(2, 0.5), (8, 0.2), (16, 0.3)])

    # --- Cenário B: binário + decimal, decimal dominante
    cenario("Sistema com I/O decimal dominante",
            [(2, 0.2), (10, 0.7), (16, 0.1)])

    # --- Cenário C: binário + decimal, binário dominante
    cenario("Sistema binário com log decimal ocasional",
            [(2, 0.7), (10, 0.2), (16, 0.1)])

    # --- Cenário D: múltiplas bases de famílias diferentes
    cenario("Protocolo heterogêneo (JSON + IP + Unix + TOTP)",
            [(10, 0.4), (16, 0.3), (8, 0.15), (32, 0.15)])

    # --- Cenário E: sistema com tempo (base 60)
    cenario("Sistema de timestamp (binário + decimal + tempo)",
            [(2, 0.4), (10, 0.4), (60, 0.2)])

    # --- Análise de sensibilidade
    print("\n" + "=" * 65)
    print("ANÁLISE DE SENSIBILIDADE: como o pivô ótimo muda")
    print("quando o peso do decimal cresce de 0% a 100%?")
    print("=" * 65)

    workload_base = [(2, 1.0), (10, 0.0)]
    resultados = analise_sensibilidade(
        workload_base, base_variavel=10,
        pesos=(0.0, 0.1, 0.25, 0.5, 0.75, 0.9, 1.0),
    )
    print(f"{'peso decimal':>14} | {'pivô ótimo':>10} | {'custo':>8}")
    print("-" * 42)
    for peso, pivo, custo, _ in resultados:
        print(f"{peso:>14.2f} | {pivo:>10} | {custo:>8.4f}")

    # --- Visualização do cenário D
    print("\nGerando gráfico do cenário D...")
    plotar_custos(
        normalizar_workload([(10, 0.4), (16, 0.3), (8, 0.15), (32, 0.15)]),
        candidatos=[2, 4, 8, 16, 32, 10, 100, 12, 60],
        titulo="Cenário D: JSON + IP + Unix + TOTP"
    )
```

```python
%%file calibrate.c
// Calibração empírica dos custos de parse e format por base.
// Saída: CSV com (base, parse_ns, format_ns).
// Este é o experimento base do Caminho A.

#include <stdio.h>
#include <stdint.h>
#include <string.h>
#include <stdlib.h>
#include <time.h>

#define N_ITER   5000000
#define N_WARMUP 200000
#define N_INPUTS 500

static inline uint64_t now_ns(void) {
    struct timespec ts;
    clock_gettime(CLOCK_MONOTONIC, &ts);
    return (uint64_t)ts.tv_sec * 1000000000ULL + ts.tv_nsec;
}

// ============ FORMATTING ============
int fmt_bin(uint64_t n, char *b){
    if(!n){b[0]='0';b[1]=0;return 1;}
    char t[66]; int i=0;
    while(n){t[i++]='0'+(n&1);n>>=1;}
    for(int j=0;j<i;j++)b[j]=t[i-1-j];
    b[i]=0;return i;
}
int fmt_oct(uint64_t n, char *b){
    if(!n){b[0]='0';b[1]=0;return 1;}
    char t[24]; int i=0;
    while(n){t[i++]='0'+(n&7);n>>=3;}
    for(int j=0;j<i;j++)b[j]=t[i-1-j];
    b[i]=0;return i;
}
int fmt_hex(uint64_t n, char *b){
    if(!n){b[0]='0';b[1]=0;return 1;}
    const char*h="0123456789abcdef";
    char t[20]; int i=0;
    while(n){t[i++]=h[n&15];n>>=4;}
    for(int j=0;j<i;j++)b[j]=t[i-1-j];
    b[i]=0;return i;
}
int fmt_dec(uint64_t n, char *b){
    if(!n){b[0]='0';b[1]=0;return 1;}
    char t[24]; int i=0;
    while(n){t[i++]='0'+n%10;n/=10;}
    for(int j=0;j<i;j++)b[j]=t[i-1-j];
    b[i]=0;return i;
}

// ============ PARSING ============
uint64_t p_bin(const char*s){uint64_t n=0;while(*s)n=(n<<1)+(*s++-'0');return n;}
uint64_t p_oct(const char*s){uint64_t n=0;while(*s)n=(n<<3)+(*s++-'0');return n;}
uint64_t p_hex(const char*s){uint64_t n=0;while(*s){char c=*s++;int d=(c>='0'&&c<='9')?c-'0':c-'a'+10;n=(n<<4)+d;}return n;}
uint64_t p_dec(const char*s){uint64_t n=0;while(*s)n=n*10+(*s++-'0');return n;}

// ============ SETUP ============
static uint64_t vals[N_INPUTS];
static char sb[N_INPUTS][80], so[N_INPUTS][24], sh[N_INPUTS][24], sd[N_INPUTS][24];

void setup(void){
    srand(42);
    for(int i=0;i<N_INPUTS;i++){
        vals[i]=((uint64_t)rand()<<32)|(uint64_t)rand();
        fmt_bin(vals[i],sb[i]);
        fmt_oct(vals[i],so[i]);
        fmt_hex(vals[i],sh[i]);
        fmt_dec(vals[i],sd[i]);
    }
}

typedef int (*fmt_fn)(uint64_t, char*);
typedef uint64_t (*parse_fn)(const char*);

double bench_parse(parse_fn pf, char **ptrs, int n){
    volatile uint64_t sink=0;
    for(int i=0;i<N_WARMUP;i++) sink += pf(ptrs[i%n]);
    uint64_t t0=now_ns();
    for(int i=0;i<N_ITER;i++) sink += pf(ptrs[i%n]);
    uint64_t dt=now_ns()-t0;
    (void)sink;
    return (double)dt/N_ITER;
}

double bench_fmt(fmt_fn ff, int n){
    char buf[128]; volatile int sink=0;
    for(int i=0;i<N_WARMUP;i++) sink += ff(vals[i%n],buf);
    uint64_t t0=now_ns();
    for(int i=0;i<N_ITER;i++) sink += ff(vals[i%n],buf);
    uint64_t dt=now_ns()-t0;
    (void)sink;
    return (double)dt/N_ITER;
}

int main(void){
    setup();
    printf("# Calibração empírica de custos de conversão\n");
    printf("# N_ITER = %d, N_INPUTS = %d\n", N_ITER, N_INPUTS);
    printf("# formato: base,parse_ns,fmt_ns\n");

    char *pb[N_INPUTS], *po[N_INPUTS], *ph[N_INPUTS], *pd[N_INPUTS];
    for(int i=0;i<N_INPUTS;i++){
        pb[i]=sb[i]; po[i]=so[i]; ph[i]=sh[i]; pd[i]=sd[i];
    }

    double t_bin = bench_parse(p_bin, pb, N_INPUTS);
    double t_oct = bench_parse(p_oct, po, N_INPUTS);
    double t_hex = bench_parse(p_hex, ph, N_INPUTS);
    double t_dec = bench_parse(p_dec, pd, N_INPUTS);

    double f_bin = bench_fmt(fmt_bin, N_INPUTS);
    double f_oct = bench_fmt(fmt_oct, N_INPUTS);
    double f_hex = bench_fmt(fmt_hex, N_INPUTS);
    double f_dec = bench_fmt(fmt_dec, N_INPUTS);

    printf("bin,%.4f,%.4f\n", t_bin, f_bin);
    printf("oct,%.4f,%.4f\n", t_oct, f_oct);
    printf("hex,%.4f,%.4f\n", t_hex, f_hex);
    printf("dec,%.4f,%.4f\n", t_dec, f_dec);

    return 0;
}
```

```python
# ============================================================
# SRCC-Decision v3: Modelo calibrado empiricamente
# ============================================================
# Substitui custos heurísticos (c_2=1, c_5=3) por custos medidos.
# Corrige três falhas do modelo anterior:
#   1. Custo proporcional ao número de dígitos, não à distância de expoentes
#   2. Conversão direcionada (parse + format), não simétrica
#   3. Pivô ótimo depende de parse+format, não do workload

import matplotlib.pyplot as plt

# --- Dados empíricos (do benchmark C) ---
PARSE_NS = {'bin': 47.28, 'oct': 18.15, 'hex': 20.80, 'dec': 17.74}
FORMAT_NS = {'bin': 89.44, 'oct': 23.13, 'hex': 19.42, 'dec': 38.39}

# --- Funções de custo ---
def parse_ns(base):
    return PARSE_NS[base]

def format_ns(base):
    return FORMAT_NS[base]

def custo_conversao_direta(origem, destino):
    """Custo de string em `origem` → string em `destino`.
    Passa por u64 internamente."""
    if origem == destino:
        return 0.0
    return PARSE_NS[origem] + FORMAT_NS[destino]

def custo_pivo(pivo):
    """Custo fixo de usar `pivo` como representação interna.
    Independente do workload."""
    return PARSE_NS[pivo] + FORMAT_NS[pivo]

# --- Ranking de pivôs ---
def ranking_pivos(bases=('bin', 'oct', 'hex', 'dec')):
    r = [(b, custo_pivo(b)) for b in bases]
    r.sort(key=lambda x: x[1])
    return r

# --- Modelo corrigido com workload ---
def custo_amortizado_corrigido(pivo, workload):
    """
    workload: lista de (base_origem, base_destino, frequência).
    Custo por transação: parse_origem + format_destino + 2*custo_pivo.
    O termo 2*custo_pivo assume que o pivô é atravessado duas vezes
    (entrada e saída).
    """
    total = 0.0
    for origem, destino, freq in workload:
        c = custo_conversao_direta(origem, destino)
        c += 2 * custo_pivo(pivo)
        total += freq * c
    return total

# ============================================================
# DEMONSTRAÇÃO
# ============================================================

print("=" * 65)
print("SRCC-Decision v3: Modelo calibrado empiricamente")
print("=" * 65)

# --- 1. Custo puro de cada pivô ---
print("\n--- Custo de usar cada base como pivô ---")
print(f"{'pivô':>6} | {'parse':>8} | {'format':>8} | {'total':>8}")
print("-" * 40)
for base, total in ranking_pivos():
    print(f"{base:>6} | {parse_ns(base):>8.2f} | {format_ns(base):>8.2f} | {total:>8.2f}")

print("\n◀ Vencedor: hexadecimal (40.22 ns), seguido de perto por octal (41.28 ns)")
print("  Binário é o PIOR pivô (136.72 ns) — 3.4x mais caro que hexa.")

# --- 2. Comparação com o modelo antigo ---
print("\n" + "=" * 65)
print("COMPARAÇÃO: modelo heurístico antigo vs. modelo calibrado")
print("=" * 65)

print("""
Modelo antigo (heurístico):
  - binário: custo 1.0 (referência)
  - decimal: custo 3.0
  - octal:   custo 1.0
  - hexa:    custo 1.0
  → Recomendava binário como pivô universal

Modelo calibrado (empírico):
  - binário: 136.72 ns  ← PIOR
  - decimal:  56.13 ns
  - octal:    41.28 ns
  - hexa:     40.22 ns  ← MELHOR
  → Recomenda hexa ou octal
""")

# --- 3. Workloads típicos, reavaliados ---
print("=" * 65)
print("WORKLOADS REAVALIADOS COM DADOS EMPÍRICOS")
print("=" * 65)

# Nota: para simplificar, assumimos que cada transação passa pelo pivô 2x.
# Isso cancela na diferença, mas mantém a interpretação correta.

cenarios = [
    ("Só bases binárias (JSON + IP + Unix + TOTP)",
     [('bin','hex',0.3), ('bin','oct',0.15), ('bin','bin',0.55)]),
    ("I/O decimal dominante",
     [('bin','dec',0.5), ('dec','bin',0.3), ('dec','dec',0.2)]),
    ("Protocolo heterogêneo",
     [('dec','hex',0.3), ('hex','oct',0.2), ('oct','dec',0.2), ('bin','dec',0.3)]),
]

for nome, workload in cenarios:
    print(f"\n--- {nome} ---")
    ranking = []
    for pivo in ['bin', 'oct', 'hex', 'dec']:
        c = custo_amortizado_corrigido(pivo, workload)
        ranking.append((pivo, c))
    ranking.sort(key=lambda x: x[1])
    for i, (p, c) in enumerate(ranking, 1):
        marcador = " ◀ ótimo" if i == 1 else ""
        print(f"  {i}. pivô {p:>4} → custo {c:>8.2f}{marcador}")

# --- 4. Visualização ---
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 5))

# Gráfico 1: parse vs format por base
bases = ['bin', 'oct', 'hex', 'dec']
x = range(len(bases))
ax1.bar([i - 0.2 for i in x], [parse_ns(b) for b in bases],
        width=0.4, label='parse', color='steelblue')
ax1.bar([i + 0.2 for i in x], [format_ns(b) for b in bases],
        width=0.4, label='format', color='coral')
ax1.set_xticks(x)
ax1.set_xticklabels(bases)
ax1.set_ylabel("nanosegundos")
ax1.set_title("Parse vs Format por base")
ax1.legend()
ax1.grid(axis='y', alpha=0.3)

# Gráfico 2: custo total por pivô
totais = [(b, custo_pivo(b)) for b in bases]
totais.sort(key=lambda x: x[1])
nomes = [t[0] for t in totais]
valores = [t[1] for t in totais]
cores = ['darkgreen' if i == 0 else 'steelblue' for i in range(len(totais))]
ax2.bar(nomes, valores, color=cores)
ax2.set_ylabel("parse + format (ns)")
ax2.set_title("Custo total por pivô (menor é melhor)")
ax2.grid(axis='y', alpha=0.3)

plt.tight_layout()
plt.show()
```

```python
%%file crossvalidate.c
// Validação cruzada: mede o pipeline A → P → B em C.
// Compara o ranking empírico com o previsto pelo modelo (parse+format).

#include <stdio.h>
#include <stdint.h>
#include <string.h>
#include <stdlib.h>
#include <time.h>

#define N_ITER   2000000
#define N_WARMUP 200000
#define N_TX     1000   // transações por workload

static inline uint64_t now_ns(void) {
    struct timespec ts;
    clock_gettime(CLOCK_MONOTONIC, &ts);
    return (uint64_t)ts.tv_sec * 1000000000ULL + ts.tv_nsec;
}

// --- Formatadores e parsers (mesmos do calibration) ---
int fmt_bin(uint64_t n, char *b){if(!n){b[0]='0';b[1]=0;return 1;}char t[66];int i=0;while(n){t[i++]='0'+(n&1);n>>=1;}for(int j=0;j<i;j++)b[j]=t[i-1-j];b[i]=0;return i;}
int fmt_oct(uint64_t n, char *b){if(!n){b[0]='0';b[1]=0;return 1;}char t[24];int i=0;while(n){t[i++]='0'+(n&7);n>>=3;}for(int j=0;j<i;j++)b[j]=t[i-1-j];b[i]=0;return i;}
int fmt_hex(uint64_t n, char *b){if(!n){b[0]='0';b[1]=0;return 1;}const char*h="0123456789abcdef";char t[20];int i=0;while(n){t[i++]=h[n&15];n>>=4;}for(int j=0;j<i;j++)b[j]=t[i-1-j];b[i]=0;return i;}
int fmt_dec(uint64_t n, char *b){if(!n){b[0]='0';b[1]=0;return 1;}char t[24];int i=0;while(n){t[i++]='0'+n%10;n/=10;}for(int j=0;j<i;j++)b[j]=t[i-1-j];b[i]=0;return i;}

uint64_t p_bin(const char*s){uint64_t n=0;while(*s)n=(n<<1)+(*s++-'0');return n;}
uint64_t p_oct(const char*s){uint64_t n=0;while(*s)n=(n<<3)+(*s++-'0');return n;}
uint64_t p_hex(const char*s){uint64_t n=0;while(*s){char c=*s++;int d=(c>='0'&&c<='9')?c-'0':c-'a'+10;n=(n<<4)+d;}return n;}
uint64_t p_dec(const char*s){uint64_t n=0;while(*s)n=n*10+(*s++-'0');return n;}

typedef int (*fmt_fn)(uint64_t, char*);
typedef uint64_t (*parse_fn)(const char*);

// Tabelas de despacho
fmt_fn   FMT[4]   = {fmt_bin, fmt_oct, fmt_hex, fmt_dec};
parse_fn PARSE[4] = {p_bin,   p_oct,   p_hex,   p_dec};
const char *NOMES[4] = {"bin", "oct", "hex", "dec"};

// --- Pipeline: string em base `src` → string em base `dst` via pivô `pivot` ---
void pipeline(const char *src_str, int src, int dst, int pivot,
              char *intermediate, char *out)
{
    // Fase 1: parse src → format pivot (armazena)
    uint64_t v = PARSE[src](src_str);
    FMT[pivot](v, intermediate);
    // Fase 2: parse pivot → format dst (emite)
    uint64_t w = PARSE[pivot](intermediate);
    FMT[dst](w, out);
}

// --- Workload: lista de (input_base, output_base) ---
typedef struct { int src, dst; } Tx;

// --- Mede o tempo de processar N_ITER vezes o workload dado um pivô ---
double bench_pivot(int pivot, Tx *txs, int n_tx,
                   const char **inputs, char *inter, char *out)
{
    volatile int sink = 0;

    // Warmup
    for (int i = 0; i < N_WARMUP; i++) {
        Tx t = txs[i % n_tx];
        pipeline(inputs[i % n_tx], t.src, t.dst, pivot, inter, out);
        sink += out[0];
    }

    uint64_t t0 = now_ns();
    for (int i = 0; i < N_ITER; i++) {
        Tx t = txs[i % n_tx];
        pipeline(inputs[i % n_tx], t.src, t.dst, pivot, inter, out);
        sink += out[0];
    }
    uint64_t dt = now_ns() - t0;
    (void)sink;
    return (double)dt / N_ITER;
}

int main(void) {
    srand(42);

    // --- Prepara 1000 strings de entrada em cada base ---
    static char str_bin[N_TX][66], str_oct[N_TX][24];
    static char str_hex[N_TX][20], str_dec[N_TX][24];
    const char *STR[4][N_TX];

    for (int i = 0; i < N_TX; i++) {
        uint64_t v = ((uint64_t)rand() << 32) | (uint64_t)rand();
        fmt_bin(v, str_bin[i]);
        fmt_oct(v, str_oct[i]);
        fmt_hex(v, str_hex[i]);
        fmt_dec(v, str_dec[i]);
    }
    for (int i = 0; i < N_TX; i++) {
        STR[0][i] = str_bin[i];
        STR[1][i] = str_oct[i];
        STR[2][i] = str_hex[i];
        STR[3][i] = str_dec[i];
    }

    // --- Três workloads ---
    Tx wl1[N_TX]; // hexa puro: todas as transações hex→hex
    Tx wl2[N_TX]; // decimal dominante
    Tx wl3[N_TX]; // heterogêneo
    for (int i = 0; i < N_TX; i++) {
        wl1[i].src = 2; wl1[i].dst = 2;
        wl2[i].src = 3; wl2[i].dst = 3;
        wl3[i].src = i % 4; wl3[i].dst = (i * 7) % 4;
    }

    // Inputs por workload: usa a base de origem do workload
    const char *in1[N_TX], *in2[N_TX], *in3[N_TX];
    for (int i = 0; i < N_TX; i++) {
        in1[i] = STR[wl1[i].src][i];
        in2[i] = STR[wl2[i].src][i];
        in3[i] = STR[wl3[i].src][i];
    }

    printf("Validação cruzada: pipeline A -> P -> B\n");
    printf("N_ITER = %d, transações = %d\n", N_ITER, N_TX);
    printf("================================================\n");

    char inter[128], out[128];

    // --- Mede cada pivô em cada workload ---
    const char *wnomes[3] = {"hexa puro", "decimal puro", "heterogêneo"};
    Tx *works[3] = {wl1, wl2, wl3};
    const char **ins[3] = {in1, in2, in3};

    for (int w = 0; w < 3; w++) {
        printf("\nWorkload: %s\n", wnomes[w]);
        printf("  %-6s | %10s | %10s | %10s | %10s\n",
               "pivô", "bin", "oct", "hex", "dec");

        double t[4];
        for (int pivot = 0; pivot < 4; pivot++) {
            t[pivot] = bench_pivot(pivot, works[w], N_TX,
                                   ins[w], inter, out);
        }
        // Ranking
        int ord[4] = {0, 1, 2, 3};
        for (int i = 0; i < 4; i++)
            for (int j = i + 1; j < 4; j++)
                if (t[ord[i]] > t[ord[j]]) {
                    int tmp = ord[i]; ord[i] = ord[j]; ord[j] = tmp;
                }

        printf("  tempos (ns/transação):\n");
        for (int i = 0; i < 4; i++) {
            printf("    %s: %.2f ns\n", NOMES[ord[i]], t[ord[i]]);
        }
    }

    return 0;
}
```

```python
%%file validate_final.c
// Validação final: rigor estatístico + varredura de magnitude + Base 32.
// Saída: CSV pronto para alimentar a figura final do artigo.

#include <stdio.h>
#include <stdint.h>
#include <string.h>
#include <stdlib.h>
#include <time.h>
#include <math.h>

#define N_ITER     1000000
#define N_WARMUP   100000
#define N_RUNS     5         // repetições para estatística
#define N_INPUTS   500

static inline uint64_t now_ns(void) {
    struct timespec ts;
    clock_gettime(CLOCK_MONOTONIC, &ts);
    return (uint64_t)ts.tv_sec * 1000000000ULL + ts.tv_nsec;
}

// ============ Formatters ============
int fmt_bin(uint64_t n, char *b){if(!n){b[0]='0';b[1]=0;return 1;}char t[66];int i=0;while(n){t[i++]='0'+(n&1);n>>=1;}for(int j=0;j<i;j++)b[j]=t[i-1-j];b[i]=0;return i;}
int fmt_oct(uint64_t n, char *b){if(!n){b[0]='0';b[1]=0;return 1;}char t[24];int i=0;while(n){t[i++]='0'+(n&7);n>>=3;}for(int j=0;j<i;j++)b[j]=t[i-1-j];b[i]=0;return i;}
int fmt_hex(uint64_t n, char *b){if(!n){b[0]='0';b[1]=0;return 1;}const char*h="0123456789abcdef";char t[20];int i=0;while(n){t[i++]=h[n&15];n>>=4;}for(int j=0;j<i;j++)b[j]=t[i-1-j];b[i]=0;return i;}
int fmt_dec(uint64_t n, char *b){if(!n){b[0]='0';b[1]=0;return 1;}char t[24];int i=0;while(n){t[i++]='0'+n%10;n/=10;}for(int j=0;j<i;j++)b[j]=t[i-1-j];b[i]=0;return i;}
// Base 32 com alfabeto RFC 4648 (A-Z, 2-7)
int fmt_b32(uint64_t n, char *b){
    const char*alpha="ABCDEFGHIJKLMNOPQRSTUVWXYZ234567";
    if(!n){b[0]='A';b[1]=0;return 1;}
    char t[16]; int i=0;
    while(n){t[i++]=alpha[n&31];n>>=5;}
    for(int j=0;j<i;j++)b[j]=t[i-1-j];
    b[i]=0;return i;
}

// ============ Parsers ============
uint64_t p_bin(const char*s){uint64_t n=0;while(*s)n=(n<<1)+(*s++-'0');return n;}
uint64_t p_oct(const char*s){uint64_t n=0;while(*s)n=(n<<3)+(*s++-'0');return n;}
uint64_t p_hex(const char*s){uint64_t n=0;while(*s){char c=*s++;int d=(c>='0'&&c<='9')?c-'0':c-'a'+10;n=(n<<4)+d;}return n;}
uint64_t p_dec(const char*s){uint64_t n=0;while(*s)n=n*10+(*s++-'0');return n;}
uint64_t p_b32(const char*s){
    uint64_t n=0;
    while(*s){
        char c=*s++;
        int d;
        if(c>='A'&&c<='Z') d=c-'A';
        else if(c>='2'&&c<='7') d=26+(c-'2');
        else d=0;
        n=(n<<5)+d;
    }
    return n;
}

typedef int (*fmt_fn)(uint64_t, char*);
typedef uint64_t (*parse_fn)(const char*);

fmt_fn   FMT[5]   = {fmt_bin, fmt_oct, fmt_hex, fmt_dec, fmt_b32};
parse_fn PARSE[5] = {p_bin,   p_oct,   p_hex,   p_dec,   p_b32};
const char *NOMES[5] = {"bin","oct","hex","dec","b32"};

// ============ Estatística ============
typedef struct { double mean; double std; } stat_t;

stat_t stats(const double *v, int n) {
    double s=0; for(int i=0;i<n;i++) s+=v[i];
    double m=s/n;
    double ss=0; for(int i=0;i<n;i++) ss+=(v[i]-m)*(v[i]-m);
    stat_t r = { m, sqrt(ss/n) };
    return r;
}

// ============ Medição com repetição ============
stat_t bench_parse(parse_fn pf, char **ptrs, int n) {
    volatile uint64_t sink=0;
    double runs[N_RUNS];
    for(int r=0;r<N_RUNS;r++){
        for(int i=0;i<N_WARMUP;i++) sink += pf(ptrs[i%n]);
        uint64_t t0=now_ns();
        for(int i=0;i<N_ITER;i++) sink += pf(ptrs[i%n]);
        uint64_t dt=now_ns()-t0;
        runs[r]=(double)dt/N_ITER;
    }
    (void)sink;
    return stats(runs, N_RUNS);
}

stat_t bench_fmt(fmt_fn ff, uint64_t *vals, int n) {
    char buf[128]; volatile int sink=0;
    double runs[N_RUNS];
    for(int r=0;r<N_RUNS;r++){
        for(int i=0;i<N_WARMUP;i++) sink += ff(vals[i%n],buf);
        uint64_t t0=now_ns();
        for(int i=0;i<N_ITER;i++) sink += ff(vals[i%n],buf);
        uint64_t dt=now_ns()-t0;
        runs[r]=(double)dt/N_ITER;
    }
    (void)sink;
    return stats(runs, N_RUNS);
}

// ============ Setup por magnitude ============
typedef enum { SMALL, MEDIUM, LARGE } magnitude_t;

uint64_t gera_valor(magnitude_t m) {
    switch(m) {
        case SMALL:  return rand() % 1000;
        case MEDIUM: return rand() % 1000000000ULL;
        case LARGE:  return ((uint64_t)rand()<<32) | (uint64_t)rand();
    }
    return 0;
}

void run_magnitude(const char *nome, magnitude_t m) {
    static uint64_t vals[N_INPUTS];
    static char buf[5][N_INPUTS][80];
    char *ptrs[5][N_INPUTS];

    srand(42); // determinístico por magnitude
    for (int i = 0; i < N_INPUTS; i++) {
        vals[i] = gera_valor(m);
        for (int b = 0; b < 5; b++) FMT[b](vals[i], buf[b][i]);
    }
    for (int i = 0; i < N_INPUTS; i++)
        for (int b = 0; b < 5; b++)
            ptrs[b][i] = buf[b][i];

    printf("# Magnitude: %s\n", nome);
    printf("# base,parse_mean_ns,parse_std_ns,fmt_mean_ns,fmt_std_ns,total_ns\n");
    for (int b = 0; b < 5; b++) {
        stat_t ps = bench_parse(PARSE[b], ptrs[b], N_INPUTS);
        stat_t fs = bench_fmt(FMT[b], vals, N_INPUTS);
        printf("%s,%.4f,%.4f,%.4f,%.4f,%.4f\n",
               NOMES[b], ps.mean, ps.std, fs.mean, fs.std,
               ps.mean + fs.mean);
    }
    printf("\n");
}

int main(void) {
    printf("# Validação final: 5 bases, 3 magnitudes, 5 repetições\n");
    printf("# N_ITER=%d, N_RUNS=%d, N_INPUTS=%d\n\n", N_ITER, N_RUNS, N_INPUTS);

    run_magnitude("small (<10^3)", SMALL);
    run_magnitude("medium (<10^9)", MEDIUM);
    run_magnitude("large (full u64)", LARGE);

    return 0;
}
```

```python
%%file crossvalidate_final.c
// Validação cruzada final: pipeline A -> P -> B com 5 bases.
// Confirma se b32 continua vencendo em pipeline real (não apenas isolado).

#include <stdio.h>
#include <stdint.h>
#include <string.h>
#include <stdlib.h>
#include <time.h>

#define N_ITER   2000000
#define N_WARMUP 200000
#define N_TX     1000

static inline uint64_t now_ns(void) {
    struct timespec ts;
    clock_gettime(CLOCK_MONOTONIC, &ts);
    return (uint64_t)ts.tv_sec * 1000000000ULL + ts.tv_nsec;
}

// ============ Formatadores ============
int fmt_bin(uint64_t n, char *b){if(!n){b[0]='0';b[1]=0;return 1;}char t[66];int i=0;while(n){t[i++]='0'+(n&1);n>>=1;}for(int j=0;j<i;j++)b[j]=t[i-1-j];b[i]=0;return i;}
int fmt_oct(uint64_t n, char *b){if(!n){b[0]='0';b[1]=0;return 1;}char t[24];int i=0;while(n){t[i++]='0'+(n&7);n>>=3;}for(int j=0;j<i;j++)b[j]=t[i-1-j];b[i]=0;return i;}
int fmt_hex(uint64_t n, char *b){if(!n){b[0]='0';b[1]=0;return 1;}const char*h="0123456789abcdef";char t[20];int i=0;while(n){t[i++]=h[n&15];n>>=4;}for(int j=0;j<i;j++)b[j]=t[i-1-j];b[i]=0;return i;}
int fmt_dec(uint64_t n, char *b){if(!n){b[0]='0';b[1]=0;return 1;}char t[24];int i=0;while(n){t[i++]='0'+n%10;n/=10;}for(int j=0;j<i;j++)b[j]=t[i-1-j];b[i]=0;return i;}
int fmt_b32(uint64_t n, char *b){
    const char*alpha="ABCDEFGHIJKLMNOPQRSTUVWXYZ234567";
    if(!n){b[0]='A';b[1]=0;return 1;}
    char t[16]; int i=0;
    while(n){t[i++]=alpha[n&31];n>>=5;}
    for(int j=0;j<i;j++)b[j]=t[i-1-j];
    b[i]=0;return i;
}

// ============ Parsers ============
uint64_t p_bin(const char*s){uint64_t n=0;while(*s)n=(n<<1)+(*s++-'0');return n;}
uint64_t p_oct(const char*s){uint64_t n=0;while(*s)n=(n<<3)+(*s++-'0');return n;}
uint64_t p_hex(const char*s){uint64_t n=0;while(*s){char c=*s++;int d=(c>='0'&&c<='9')?c-'0':c-'a'+10;n=(n<<4)+d;}return n;}
uint64_t p_dec(const char*s){uint64_t n=0;while(*s)n=n*10+(*s++-'0');return n;}
uint64_t p_b32(const char*s){
    uint64_t n=0;
    while(*s){
        char c=*s++;
        int d;
        if(c>='A'&&c<='Z') d=c-'A';
        else if(c>='2'&&c<='7') d=26+(c-'2');
        else d=0;
        n=(n<<5)+d;
    }
    return n;
}

typedef int (*fmt_fn)(uint64_t, char*);
typedef uint64_t (*parse_fn)(const char*);

#define NBASES 5
fmt_fn   FMT[NBASES]   = {fmt_bin, fmt_oct, fmt_hex, fmt_dec, fmt_b32};
parse_fn PARSE[NBASES] = {p_bin,   p_oct,   p_hex,   p_dec,   p_b32};
const char *NOMES[NBASES] = {"bin","oct","hex","dec","b32"};

// ============ Pipeline A -> P -> B ============
void pipeline(const char *src_str, int src, int dst, int pivot,
              char *intermediate, char *out)
{
    uint64_t v = PARSE[src](src_str);
    FMT[pivot](v, intermediate);
    uint64_t w = PARSE[pivot](intermediate);
    FMT[dst](w, out);
}

typedef struct { int src, dst; } Tx;

double bench_pivot(int pivot, Tx *txs, int n_tx,
                   const char **inputs, char *inter, char *out)
{
    volatile int sink = 0;
    for (int i = 0; i < N_WARMUP; i++) {
        Tx t = txs[i % n_tx];
        pipeline(inputs[i % n_tx], t.src, t.dst, pivot, inter, out);
        sink += out[0];
    }
    uint64_t t0 = now_ns();
    for (int i = 0; i < N_ITER; i++) {
        Tx t = txs[i % n_tx];
        pipeline(inputs[i % n_tx], t.src, t.dst, pivot, inter, out);
        sink += out[0];
    }
    uint64_t dt = now_ns() - t0;
    (void)sink;
    return (double)dt / N_ITER;
}

// ============ Setup ============
static char str_all[NBASES][N_TX][80];
const char *STR[NBASES][N_TX];

void prepara_strings(void) {
    srand(42);
    for (int i = 0; i < N_TX; i++) {
        uint64_t v = ((uint64_t)rand()<<32) | (uint64_t)rand();
        for (int b = 0; b < NBASES; b++)
            FMT[b](v, str_all[b][i]);
    }
    for (int i = 0; i < N_TX; i++)
        for (int b = 0; b < NBASES; b++)
            STR[b][i] = str_all[b][i];
}

void run_workload(const char *nome, Tx *txs, int n_tx) {
    char inter[128], out[128];
    static const char *ins[N_TX];
    for (int i = 0; i < n_tx; i++)
        ins[i] = STR[txs[i].src][i];

    printf("\n=== Workload: %s ===\n", nome);
    double t[NBASES];
    for (int p = 0; p < NBASES; p++)
        t[p] = bench_pivot(p, txs, n_tx, ins, inter, out);

    // Ranking
    int ord[NBASES];
    for (int i = 0; i < NBASES; i++) ord[i] = i;
    for (int i = 0; i < NBASES; i++)
        for (int j = i+1; j < NBASES; j++)
            if (t[ord[i]] > t[ord[j]]) {
                int tmp = ord[i]; ord[i] = ord[j]; ord[j] = tmp;
            }

    printf("  Ranking (menor = melhor):\n");
    for (int i = 0; i < NBASES; i++) {
        const char *marca = (i == 0) ? "  <-- OTIMO" : "";
        printf("    %d. %-4s %8.2f ns/tx%s\n",
               i+1, NOMES[ord[i]], t[ord[i]], marca);
    }
}

int main(void) {
    prepara_strings();

    printf("Crossvalidation FINAL: 5 bases, 3 workloads\n");
    printf("N_ITER=%d, N_TX=%d\n", N_ITER, N_TX);
    printf("==============================================\n");

    // Workload 1: "hexa puro" (só hex→hex)
    static Tx wl1[N_TX];
    for (int i = 0; i < N_TX; i++) { wl1[i].src = 2; wl1[i].dst = 2; }
    run_workload("hexa puro", wl1, N_TX);

    // Workload 2: "decimal puro"
    static Tx wl2[N_TX];
    for (int i = 0; i < N_TX; i++) { wl2[i].src = 3; wl2[i].dst = 3; }
    run_workload("decimal puro", wl2, N_TX);

    // Workload 3: "heterogêneo" (todos com todos)
    static Tx wl3[N_TX];
    for (int i = 0; i < N_TX; i++) {
        wl3[i].src = i % NBASES;
        wl3[i].dst = (i * 7) % NBASES;
    }
    run_workload("heterogêneo", wl3, N_TX);

    // Workload 4: "real-world" (JSON 40%, IPv6 30%, Unix 15%, TOTP 15%)
    // Codificações: JSON = dec, IPv6 = hex, Unix = oct, TOTP = b32
    static Tx wl4[N_TX];
    for (int i = 0; i < N_TX; i++) {
        int r = i % 100;
        int base_src;
        if (r < 40) base_src = 3;       // dec
        else if (r < 70) base_src = 2;  // hex
        else if (r < 85) base_src = 1;  // oct
        else base_src = 4;              // b32
        // Destino: sempre para decimal (consolidação)
        wl4[i].src = base_src;
        wl4[i].dst = 3;
    }
    run_workload("real-world (JSON+IPv6+Unix+TOTP)", wl4, N_TX);

    return 0;
}
```

```python
%%file sanity_workload.c
// Workload de sanidade: mix realista de formatos de produção.
// Modela três fontes de dados com distribuições distintas:
//   1. JSON de API     - dominado por decimal
//   2. XML de config   - mistura decimal + hex (atributos ID)
//   3. Logs de sistema - dominado por hex (timestamps, endereços)
// Mais uma workload sintética "caos controlado" com todas as bases.
//
// O objetivo é confirmar se b32 continua vencendo em cenário realista.

#include <stdio.h>
#include <stdint.h>
#include <string.h>
#include <stdlib.h>
#include <time.h>

#define N_ITER   2000000
#define N_WARMUP 200000
#define N_TX     1000

static inline uint64_t now_ns(void) {
    struct timespec ts;
    clock_gettime(CLOCK_MONOTONIC, &ts);
    return (uint64_t)ts.tv_sec * 1000000000ULL + ts.tv_nsec;
}

// ============ Formatadores ============
int fmt_bin(uint64_t n, char *b){if(!n){b[0]='0';b[1]=0;return 1;}char t[66];int i=0;while(n){t[i++]='0'+(n&1);n>>=1;}for(int j=0;j<i;j++)b[j]=t[i-1-j];b[i]=0;return i;}
int fmt_oct(uint64_t n, char *b){if(!n){b[0]='0';b[1]=0;return 1;}char t[24];int i=0;while(n){t[i++]='0'+(n&7);n>>=3;}for(int j=0;j<i;j++)b[j]=t[i-1-j];b[i]=0;return i;}
int fmt_hex(uint64_t n, char *b){if(!n){b[0]='0';b[1]=0;return 1;}const char*h="0123456789abcdef";char t[20];int i=0;while(n){t[i++]=h[n&15];n>>=4;}for(int j=0;j<i;j++)b[j]=t[i-1-j];b[i]=0;return i;}
int fmt_dec(uint64_t n, char *b){if(!n){b[0]='0';b[1]=0;return 1;}char t[24];int i=0;while(n){t[i++]='0'+n%10;n/=10;}for(int j=0;j<i;j++)b[j]=t[i-1-j];b[i]=0;return i;}
int fmt_b32(uint64_t n, char *b){
    const char*alpha="ABCDEFGHIJKLMNOPQRSTUVWXYZ234567";
    if(!n){b[0]='A';b[1]=0;return 1;}
    char t[16]; int i=0;
    while(n){t[i++]=alpha[n&31];n>>=5;}
    for(int j=0;j<i;j++)b[j]=t[i-1-j];
    b[i]=0;return i;
}

// ============ Parsers ============
uint64_t p_bin(const char*s){uint64_t n=0;while(*s)n=(n<<1)+(*s++-'0');return n;}
uint64_t p_oct(const char*s){uint64_t n=0;while(*s)n=(n<<3)+(*s++-'0');return n;}
uint64_t p_hex(const char*s){uint64_t n=0;while(*s){char c=*s++;int d=(c>='0'&&c<='9')?c-'0':c-'a'+10;n=(n<<4)+d;}return n;}
uint64_t p_dec(const char*s){uint64_t n=0;while(*s)n=n*10+(*s++-'0');return n;}
uint64_t p_b32(const char*s){
    uint64_t n=0;
    while(*s){
        char c=*s++;
        int d;
        if(c>='A'&&c<='Z') d=c-'A';
        else if(c>='2'&&c<='7') d=26+(c-'2');
        else d=0;
        n=(n<<5)+d;
    }
    return n;
}

typedef int (*fmt_fn)(uint64_t, char*);
typedef uint64_t (*parse_fn)(const char*);

#define NBASES 5
fmt_fn   FMT[NBASES]   = {fmt_bin, fmt_oct, fmt_hex, fmt_dec, fmt_b32};
parse_fn PARSE[NBASES] = {p_bin,   p_oct,   p_hex,   p_dec,   p_b32};
const char *NOMES[NBASES] = {"bin","oct","hex","dec","b32"};

void pipeline(const char *src_str, int src, int dst, int pivot,
              char *intermediate, char *out)
{
    uint64_t v = PARSE[src](src_str);
    FMT[pivot](v, intermediate);
    uint64_t w = PARSE[pivot](intermediate);
    FMT[dst](w, out);
}

typedef struct { int src, dst; } Tx;

double bench_pivot(int pivot, Tx *txs, int n_tx,
                   const char **inputs, char *inter, char *out)
{
    volatile int sink = 0;
    for (int i = 0; i < N_WARMUP; i++) {
        Tx t = txs[i % n_tx];
        pipeline(inputs[i % n_tx], t.src, t.dst, pivot, inter, out);
        sink += out[0];
    }
    uint64_t t0 = now_ns();
    for (int i = 0; i < N_ITER; i++) {
        Tx t = txs[i % n_tx];
        pipeline(inputs[i % n_tx], t.src, t.dst, pivot, inter, out);
        sink += out[0];
    }
    uint64_t dt = now_ns() - t0;
    (void)sink;
    return (double)dt / N_ITER;
}

static char str_all[NBASES][N_TX][80];
const char *STR[NBASES][N_TX];

void prepara_strings(void) {
    srand(42);
    for (int i = 0; i < N_TX; i++) {
        uint64_t v = ((uint64_t)rand()<<32) | (uint64_t)rand();
        for (int b = 0; b < NBASES; b++)
            FMT[b](v, str_all[b][i]);
    }
    for (int i = 0; i < N_TX; i++)
        for (int b = 0; b < NBASES; b++)
            STR[b][i] = str_all[b][i];
}

void run_workload(const char *nome, Tx *txs, int n_tx) {
    char inter[128], out[128];
    static const char *ins[N_TX];
    for (int i = 0; i < n_tx; i++)
        ins[i] = STR[txs[i].src][i];

    printf("\n=== Workload: %s ===\n", nome);
    double t[NBASES];
    for (int p = 0; p < NBASES; p++)
        t[p] = bench_pivot(p, txs, n_tx, ins, inter, out);

    int ord[NBASES];
    for (int i = 0; i < NBASES; i++) ord[i] = i;
    for (int i = 0; i < NBASES; i++)
        for (int j = i+1; j < NBASES; j++)
            if (t[ord[i]] > t[ord[j]]) {
                int tmp = ord[i]; ord[i] = ord[j]; ord[j] = tmp;
            }

    printf("  Ranking (menor = melhor):\n");
    for (int i = 0; i < NBASES; i++) {
        const char *marca = (i == 0) ? "  <-- OTIMO" : "";
        printf("    %d. %-4s %8.2f ns/tx%s\n",
               i+1, NOMES[ord[i]], t[ord[i]], marca);
    }
}

// ============ Geração de workloads ============

// Workload 1: JSON API
// - 70% decimal (números em payload)
// - 20% hex (hashes de request, IDs)
// - 10% b32 (tokens de sessão)
// Destino: JSON response, sempre decimal
void gera_json(Tx *txs, int n) {
    for (int i = 0; i < n; i++) {
        int r = i % 100;
        int src;
        if (r < 70) src = 3;        // dec
        else if (r < 90) src = 2;   // hex
        else src = 4;               // b32
        txs[i].src = src;
        txs[i].dst = 3;             // sempre responde em JSON (dec)
    }
}

// Workload 2: XML de configuração
// - 50% decimal (valores numéricos)
// - 30% hex (IDs de elemento)
// - 20% oct (modo de arquivo Unix)
// Destino: mistura XML (dec) e logs estruturados (hex)
void gera_xml(Tx *txs, int n) {
    for (int i = 0; i < n; i++) {
        int r = i % 100;
        int src;
        if (r < 50) src = 3;        // dec
        else if (r < 80) src = 2;   // hex
        else src = 1;               // oct
        txs[i].src = src;
        txs[i].dst = (i % 2 == 0) ? 3 : 2;  // dec ou hex
    }
}

// Workload 3: Logs de sistema
// - 60% hex (timestamps, endereços, hashes)
// - 30% decimal (métricas)
// - 10% b32 (trace IDs)
// Destino: formato binário compacto (bin) ou hex
void gera_logs(Tx *txs, int n) {
    for (int i = 0; i < n; i++) {
        int r = i % 100;
        int src;
        if (r < 60) src = 2;        // hex
        else if (r < 90) src = 3;   // dec
        else src = 4;               // b32
        txs[i].src = src;
        txs[i].dst = (i % 3 == 0) ? 0 : 2;  // bin ou hex
    }
}

// Workload 4: sistema em produção completo
// Agrega as três fontes acima com pesos realistas:
// - 50% JSON, 30% XML, 20% logs
void gera_producao(Tx *txs, int n) {
    for (int i = 0; i < n; i++) {
        int r = i % 100;
        if (r < 50) {
            // JSON slice
            int rr = (i * 3) % 100;
            int src;
            if (rr < 70) src = 3;
            else if (rr < 90) src = 2;
            else src = 4;
            txs[i].src = src;
            txs[i].dst = 3;
        } else if (r < 80) {
            // XML slice
            int rr = (i * 7) % 100;
            int src;
            if (rr < 50) src = 3;
            else if (rr < 80) src = 2;
            else src = 1;
            txs[i].src = src;
            txs[i].dst = (i % 2 == 0) ? 3 : 2;
        } else {
            // Logs slice
            int rr = (i * 11) % 100;
            int src;
            if (rr < 60) src = 2;
            else if (rr < 90) src = 3;
            else src = 4;
            txs[i].src = src;
            txs[i].dst = (i % 3 == 0) ? 0 : 2;
        }
    }
}

int main(void) {
    prepara_strings();

    printf("SANITY: workloads realistas de produção\n");
    printf("N_ITER=%d, N_TX=%d\n", N_ITER, N_TX);
    printf("==============================================\n");

    static Tx wl[N_TX];

    gera_json(wl, N_TX);
    run_workload("JSON API (70%% dec, 20%% hex, 10%% b32)", wl, N_TX);

    gera_xml(wl, N_TX);
    run_workload("XML config (50%% dec, 30%% hex, 20%% oct)", wl, N_TX);

    gera_logs(wl, N_TX);
    run_workload("System logs (60%% hex, 30%% dec, 10%% b32)", wl, N_TX);

    gera_producao(wl, N_TX);
    run_workload("PRODUÇÃO COMPLETA (50%% JSON + 30%% XML + 20%% logs)", wl, N_TX);

    return 0;
}
```
